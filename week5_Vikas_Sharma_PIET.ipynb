```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, GRU, Dense
import numpy as np
import matplotlib.pyplot as plt
print("TensorFlow:", tf.__version__)
```

# 📥 Load Text Corpus
We use a **small built-in sample corpus** so students can run this quickly.

- chatbot data

- custom PDF extracted text

```python
corpus = '''
Artificial intelligence is transforming industries across the world.
Machine learning enables computers to learn patterns from data.
Deep learning uses multiple neural network layers to solve complex problems.
Neural networks are inspired by the structure of the human brain.
Convolutional neural networks are widely used for image recognition tasks.
Recurrent neural networks are designed to process sequential information.
Long Short-Term Memory networks help retain important information over long sequences.
Gated Recurrent Units provide a simpler and faster alternative to LSTM networks.
Transformers have revolutionized natural language processing and text generation.
Attention mechanisms allow models to focus on relevant parts of the input.
Large language models can generate human-like text and answer questions.
Deep learning models require large datasets for effective training.
Data preprocessing improves the quality and performance of machine learning models.
Gradient descent is a popular optimization algorithm used in training neural networks.
Backpropagation updates model parameters by minimizing prediction errors.
Overfitting occurs when a model memorizes training data instead of learning patterns.
Regularization techniques help improve model generalization.
Dropout reduces overfitting by randomly disabling neurons during training.
Transfer learning allows models to reuse knowledge from previously learned tasks.
Computer vision enables machines to understand and analyze visual information.
Speech recognition systems convert spoken language into text.
Natural language processing helps computers understand human language.
Generative AI can create text, images, audio, and video content.
Text generation models predict the most likely next word in a sequence.
Recommendation systems suggest relevant products and content to users.
Autonomous vehicles use deep learning for navigation and object detection.
Reinforcement learning trains agents through rewards and penalties.
AI-powered chatbots assist users with information and customer support.
Deep learning is driving innovation in healthcare, finance, and education.
Modern AI systems continue to improve through research and advanced computing resources.
Neural network architectures are becoming more efficient and scalable.
Explainable AI aims to make machine learning decisions more transparent.
Federated learning enables training across multiple devices while preserving privacy.
Self-supervised learning reduces dependence on manually labeled datasets.
Multimodal AI models can process text, images, audio, and video simultaneously.
The future of artificial intelligence promises smarter and more adaptive systems.
'''
print(corpus)
```

# 🔤 Tokenization & Sequence Creation
We convert text into integer tokens and create **n-gram style sequences**
for next-word prediction.

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([corpus])

total_words = len(tokenizer.word_index) + 1
print("Vocabulary size:", total_words)

input_sequences = []
for line in corpus.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_seq = token_list[:i+1]
        input_sequences.append(n_gram_seq)

max_len = max(len(seq) for seq in input_sequences)
input_sequences = pad_sequences(input_sequences, maxlen=max_len, padding='pre')

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

# 🧠 Model 1: Vanilla RNN
This is the baseline sequential model.
It struggles with long-term dependencies because of vanishing gradients.

```python
rnn_model = Sequential([
    Embedding(total_words, 64, input_length=max_len-1),
    SimpleRNN(128),
    Dense(total_words, activation='softmax')
])

rnn_model.compile(loss='sparse_categorical_crossentropy',
                  optimizer='adam',
                  metrics=['accuracy'])

rnn_history = rnn_model.fit(X, y, epochs=200, verbose=0)
print("Vanilla RNN training completed")
```

# 🔒 Model 2: LSTM
LSTM uses **input, forget, and output gates**
to preserve long-term memory.

```python
lstm_model = Sequential([
    Embedding(total_words, 64, input_length=max_len-1),
    LSTM(128),
    Dense(total_words, activation='softmax')
])

lstm_model.compile(loss='sparse_categorical_crossentropy',
                   optimizer='adam',
                   metrics=['accuracy'])

lstm_history = lstm_model.fit(X, y, epochs=200, verbose=0)
print("LSTM training completed")
```

# ⚡ Model 3: GRU
GRU uses **reset + update gates**.
It is computationally faster than LSTM and often gives similar results.

```python
gru_model = Sequential([
    Embedding(total_words, 64, input_length=max_len-1),
    GRU(128),
    Dense(total_words, activation='softmax')
])

gru_model.compile(loss='sparse_categorical_crossentropy',
                  optimizer='adam',
                  metrics=['accuracy'])

gru_history = gru_model.fit(X, y, epochs=200, verbose=0)
print("GRU training completed")
```

## 📉 Compare Training Loss

```python
plt.figure(figsize=(10,4))
plt.plot(rnn_history.history['loss'], label='RNN')
plt.plot(lstm_history.history['loss'], label='LSTM')
plt.plot(gru_history.history['loss'], label='GRU')
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Comparison")
plt.legend()
plt.show()
```

# ✍️ Text Generation Function
This function predicts the next word repeatedly to generate a sentence.

```python
def generate_text(model, seed_text, next_words=10):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences([token_list], maxlen=max_len-1, padding='pre')
        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)[0]

        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break
        seed_text += " " + output_word
    return seed_text
```

## 🧪 Generate Text Samples

```python
print("RNN :", generate_text(rnn_model, "deep learning", 10))
print("LSTM:", generate_text(lstm_model, "deep learning", 10))
print("GRU :", generate_text(gru_model, "deep learning", 10))
```
